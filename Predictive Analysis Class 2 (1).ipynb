```python
df.Name.str.extract('([A-Za-z]+)\.')
```

```python
df['Initial'] = 0
for i in df.Name:
    df['Initial']=df.Name.str.extract('([A-Za-z]+)\.') #lets extract the Salutations
```

```python
X_test['Initial']=0
for i in X_test.Name:
    X_test['Initial']=X_test.Name.str.extract('([A-Za-z]+)\.') #lets extract the Salutations
```
